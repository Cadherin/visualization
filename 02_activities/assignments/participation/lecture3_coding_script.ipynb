```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1)
ax.plot(x,y2)
fig.show()


```

```python
# Add labels and legend

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')
fig.show()
```

```python
# Modify legend

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = 'lower right',
          frameon = True, # add frame around legend
          fontsize = 12, # update font size of legend
          ncol = 2, # Make legend to display in two columns
          shadow = True # Create shadow around the legend frame
          )
fig.show()
```

```python
# Move legend outside of plot area

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = 'upper left',
          bbox_to_anchor = (1,1),
          frameon = True, # add frame around legend
          fontsize = 12, # update font size of legend
          ncol = 2, # Make legend to display in two columns
          shadow = True # Create shadow around the legend frame
          )
fig.show()

```

```python
# Move legend outside of plot area

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = 'upper left',
          bbox_to_anchor = (1,1),
          frameon = False, # add frame around legend
          fontsize = 12, # update font size of legend
          ncol = 1, # Make legend to display in two columns
          shadow = True # Create shadow around the legend frame
          )
fig.show()
```

```python
# Modify text

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right',
          #bbox_to_anchor = (1,1),
          frameon = True, # add frame around legend
          fontsize = 12, # update font size of legend
          ncol = 1, # Make legend to display in two columns
          shadow = True # Create shadow around the legend frame
          )
ax.text(10,95, "This value is important!")
fig.show()
```

```python
# Further modify text

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right',
          #bbox_to_anchor = (1,1),
          frameon = True, # add frame around legend
          fontsize = 12, # update font size of legend
          ncol = 1, # Make legend to display in two columns
          shadow = True # Create shadow around the legend frame
          )
ax.text(10,95, "This value is important!",
        ha = 'center',
        color = 'red',
        size = 20)
fig.show()
```

```python
# Text annotations
fig, ax= plt.subplots()
ax.axis([0,10,0,10])
ax.text(1,5,".Data:(1,5)", transform=ax.transData)
ax.text(0.5,0.1,".Axes:(0.5,0.1)", transform=ax.transAxes)
ax.text(0.2,0.2,".Figure:(0.2,0.2)", transform=fig.transFigure)
```

```python
# Annotate with arrows
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10,95),xytext=(20,94), arrowprops=dict(facecolor='black'))
```

```python
# Stylize arrow
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94), arrowprops = dict(arrowstyle = "wedge", color = "hotpink"))
```

```python
# Remove tick marks or labels on x- or y-axis
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.yaxis.set_major_locator(plt.NullLocator()) 
ax.xaxis.set_major_formatter(plt.NullFormatter())
```

```python
# Control total number of tick marks on x-axis
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
# Rotate x-axis labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xticks(rotation=45, ha='right')
```
