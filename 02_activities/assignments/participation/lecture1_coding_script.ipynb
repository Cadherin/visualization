```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Generate data points for x, y

np.random.seed(613) # set random seed for reproducibility
x = np.arange(50) # create 50 integers from 0 to 49
y = np.random.randint(0,100,50) # create 50 random integers taken between 0 and 100  
```

```python
# Plot x, y values in a scatter plot

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y)
```

```python
# Plot x, y values in a bar chart

fig, ax = plt.subplots(figsize=(5,3))
ax.bar(x,y)
```

```python
# Plot x, y values in a line chart

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
```

```python
# Create a histogram

fig, ax = plt.subplots(figsize=(5,3))
ax.hist(y)
```

```python
# Adding axis labels and titles to the above plot

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total grwoth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
# Modify chart labels and titles

# Define fonts
font1={'family':'sans-serif','color':'blue', 'size':20}
font2={'family':'monospace','color':'green', 'size':14}

# Create plot and enhance fonts of title and labels
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict=font1, loc='right')
ax.set_ylabel('Total grwoth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)
fig.tight_layout()

```

```python
# Change color and marker for scatter plot

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y,
           marker = '*',
           color = 'indigo'
           )
```

```python
# Add line to scatter plot, need to use a line plot

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y,
           marker = '*',
           color = 'indigo',
           linestyle = '--',
           linewidth = 2
           )
```

```python
# Further customize plot by changing marker color

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y,
           marker = '*',
           color = '#7425b9',
           linestyle = '--',
           linewidth = 2,
           markersize = 10,
           markeredgecolor = '#fa9359',
           markerfacecolor = '#000000'
           )
```

```python
# Add grid line to chart

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y,
           marker = '*',
           color = '#7425b9',
           linestyle = '--',
           linewidth = 2,
           markersize = 10,
           markeredgecolor = '#fa9359',
           markerfacecolor = '#000000'
           )

ax.grid(axis = 'x', 
        color = 'blue', 
        linewidth = 1, 
        linestyle = '--'
)

ax.grid(axis = 'y', 
        color = 'blue', 
        linewidth = 1, 
        linestyle = '--'
)
```
