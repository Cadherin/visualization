### Accessibility of visualization

1. High contrast generally makes visualization easy to consume. Brightness that is too high can be difficult for some people to view. 

2. Sarah L. Fossheim, a developer and designer, created different variations of the the daily routines for famous creative people visualization using different color schemes to explore what people with different types of color blindness actually see. It was found that the original color chosen may not be accessible to some people. 

3. Color has 3 channels: 
    - Luminance: how bright
    - Saturation: how colourful
    - Hue: pure colors (suitable for categorical data, identity channel)


4. Alt text should be incorporated to ensure that people using a screen reader or unable to see the image could still understand what is being shown. Alt text should be in plain language and simple sentences as much as possible.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Generate dataset

np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
# Create a canvas for the figure
fig, ax = plt.subplots(figsize=(5, 3))
```

```python
# Create two plots next to each other, we call these the subplots

fig,(ax1, ax2) = plt.subplots(ncols=2, 
                              nrows=1, 
                              figsize=(7, 3))
```

```python
# Plot data into the subplots
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
fig.show()
```

```python
# Add titles to subplots
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
ax1.set_title('Scatter Plot')
ax2.set_title('Bar Chart')
```

```python
# Show styles available
plt.style.available
```

```python
# Update chart color to dark palette
# Update title font to serif and title color to blue
# Update marker shape of the scatter plot to a star

plt.style.use('seaborn-v0_8-dark-palette')
font1 = {'family':'serif','color':'blue'}
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))
ax1.scatter(x1,y1, marker = '*')
ax2.bar(x2,y2)
ax1.set_title('Scatter Plot', fontdict = font1)
ax2.set_title('Bar Chart', fontdict = font1)
```

```python
# Use subplot mosaic to change grid arrangement

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize=(7, 4))
```

```python
# Create subplots with mosaic arrangement 

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
plt.show()
```

```python
# Add very large x-axis labels

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], 
                                    ['ax2', 'ax3']], 
                                    figsize=(7, 4)) 

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)

someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18)
```

```python
# Apply contrained layout


fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], 
                                    ['ax2', 'ax3']], 
                                    figsize=(7, 4),
                                    layout = 'constrained') 

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)

someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18)
```

```python
# Create multiple plots on the same axes

# first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))

# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,color = "indigo")
ax.plot(x, y2,color = "red")
```

```python
# Add error information

# Calculate standard deviation
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")

# Error information
ax.errorbar(x, #our x values 
            y2, #our y values 
            yerr = y2_sd, 
            fmt = "none")
```

```python
# Customize error bar appearance

fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax.errorbar(x, y2, 
            yerr = y2_sd, 
            fmt = "none", 
            ecolor= "indigo", 
            elinewidth= 4, 
            capsize = 6, 
            capthick= 4 )
```

```python
# Use errorevery to specify how often the error bar appears
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax.errorbar(x, y2, 
            yerr = y2_sd, 
            fmt = "none", 
            ecolor = "indigo", 
            elinewidth = 4, 
            capsize = 6, 
            capthick = 4, 
            errorevery= 2)
```

### Setup for using images

```python
from PIL import Image # to open images
import requests # to get images from URLs
from io import BytesIO # to store images
```

```python
# Get image from internet

url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"
headers = {"User-Agent": "Mozilla/5.0","Accept": "image/*,*/*;q=0.8","Referer": "https://en.wikipedia.org/"}
response = requests.get(url, headers=headers)
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
# Show image
image
```

```python
# Make a simple plot
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")

# Overlay image on top of plot
ax_image = fig.add_axes([0.1, # x coordinate (ON FIGURE, NOT AXES) 
                         0.11, # y coordinate (ON FIGURE, NOT AXES) 
                         0.15, # image width 
                         0.35] # image height
                         )

ax_image.imshow(image)
ax_image.axis('off')
```
